## Latest comments for a stock

Codes work with or without the `.T` market suffix.


````python
from yfcomment import get_forum

forum = get_forum("285A", limit=10)   # Kioxia
forum.poll
````

````python
for c in forum.comments[:5]:
    print(f"{c.posted_at:%m/%d %H:%M}  +{c.helpful_yes}/-{c.helpful_no}  {c.text[:60]}")
````

## As a DataFrame


````python
import pandas as pd

df = pd.DataFrame(c.to_dict() for c in forum.comments)
df.head()
````

## The comment ranking

Which stocks is everyone talking about? `term` is `"daily"`, `"weekly"`, or `"monthly"`.


````python
from yfcomment import get_ranking

pd.DataFrame(e.to_dict() for e in get_ranking("daily", limit=15))
````

## CLI

The same data is available in your terminal:

```bash
yfc 285A -n 50        # comments table
yfc rank -w           # weekly ranking
yfc 285A --json | jq  # scriptable
```
